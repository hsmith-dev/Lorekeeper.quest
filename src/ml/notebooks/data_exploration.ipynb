# Lorekeeper — Training Data Exploration

Explores the generated training data quality before fine-tuning.

```python
import json
from pathlib import Path

DATA_DIR = Path('../data')
RAW_DIR = DATA_DIR / 'raw'
PROCESSED_DIR = DATA_DIR / 'processed'
```

```python
def load_jsonl(path):
    records = []
    with open(path) as f:
        for line in f:
            if line.strip():
                records.append(json.loads(line))
    return records

# Load all raw data
all_raw = {}
for path in sorted(RAW_DIR.glob('*.jsonl')):
    genre = path.stem.replace('_samples', '')
    all_raw[genre] = load_jsonl(path)
    print(f'{genre}: {len(all_raw[genre])} records')
```

```python
# Check output completion rate
for genre, records in all_raw.items():
    completed = sum(1 for r in records if r.get('output', '').strip())
    print(f'{genre}: {completed}/{len(records)} outputs filled ({100*completed//max(len(records),1)}%)')
```

```python
# Output length distribution
import statistics

for genre, records in all_raw.items():
    lengths = [len(r.get('output', '')) for r in records if r.get('output', '').strip()]
    if lengths:
        print(f'{genre}: min={min(lengths)} max={max(lengths)} median={statistics.median(lengths):.0f}')
```

```python
# Sample a few examples
import random

genre = 'dnd'
records = [r for r in all_raw.get(genre, []) if r.get('output', '').strip()]
if records:
    sample = random.choice(records)
    print('INPUT:', sample['input'])
    print()
    print('OUTPUT:', sample['output'])
```
